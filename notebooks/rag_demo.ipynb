# RAG Document Q&A Demo

This notebook demonstrates the API flow after the FastAPI server is running locally.

```python
import requests

BASE_URL = "http://127.0.0.1:8000"
```

```python
sample_text = """
Retrieval-Augmented Generation combines search with text generation.
The retriever finds relevant document chunks, and the generator answers using those chunks.
Grounded answers should avoid using facts that are not present in the retrieved context.
""".strip()

with open("sample_rag_note.txt", "w", encoding="utf-8") as file:
    file.write(sample_text)

with open("sample_rag_note.txt", "rb") as file:
    response = requests.post(BASE_URL + "/ingest", files={"file": ("sample_rag_note.txt", file)})

response.raise_for_status()
doc = response.json()
doc
```

```python
query = {
    "doc_id": doc["doc_id"],
    "question": "What does the retriever do?",
    "top_k": 3,
}
response = requests.post(BASE_URL + "/query", json=query)
response.raise_for_status()
response.json()
```
